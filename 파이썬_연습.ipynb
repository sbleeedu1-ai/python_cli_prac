```python
from pytz import timezone
from datetime import datetime
today = datetime.now(timezone('Asia/Seoul'))


article_num = 0
articles = []

print("== CLI 게시판 실행 ==")

while True:
  cmd = input("명령어 ) ")

  if cmd == 'article write':
    article_num += 1
    title = input("제목 : ")
    body = input("내용 : ")
    write_time = today.strftime('%Y-%m-%d %H:%M:%S')
    edit_time = today.strftime('%Y-%m-%d %H:%M:%S')
    article = [article_num, title, body, write_time, edit_time]

    articles.append(article)
    print(f"{article_num} 번 글이 생성 되었습니다.")

  elif cmd == 'article list':
    print("=======================================")
    print("번호".ljust(6)+"/",end='')
    print("제목".center(13)+"/",end='')
    print("내용".center(13)+"/",end='')
    print("작성시간".rjust(8))
    for i in range(len(articles)-1,-1,-1):
      print(f"{articles[i][0]}".ljust(8)+"/",end='')
      print(f"{articles[i][1]}".center(15)+"/",end='')
      print(f"{articles[i][2]}".center(15)+"/",end='')
      print(f"    {articles[i][3][:10]}")
    print("=======================================")

  elif cmd.startswith('article delete'):
    cmd_bit = cmd[14:].strip()

    if cmd_bit == '':
      print("입력된 글 번호가 없습니다.")
      continue

    if not cmd_bit.isdigit():
      print("글 번호가 아닙니다.")
      continue

    article_exist = False
    for i in range(0,len(articles)):
      if articles[i][0] == int(cmd_bit):
        a = articles[i]
        article_exist = True
        break

    if article_exist == False:
      print(f"{cmd_bit} 번 글이 존재하지 않습니다.")
      continue

    articles.remove(a)
    print(f"{cmd_bit} 번 글이 삭제 되었습니다.")

  elif cmd.startswith('article edit'):
    cmd_bit = cmd[12:].strip()

    if cmd_bit == '':
      print("입력된 글 번호가 없습니다.")
      continue

    if not cmd_bit.isdigit():
      print("글 번호가 아닙니다.")
      continue

    article_exist = False
    for i in range(0,len(articles)):
      if articles[i][0] == int(cmd_bit):
        a = articles[i]
        article_exist = True
        break

    if article_exist == False:
      print(f"{cmd_bit} 번 글이 존재하지 않습니다.")
      continue

    print("기존 제목 : {}".format(a[1]))
    print("기존 내용 : {}".format(a[2]))
    a[1] = input("새 제목 : ")
    a[2] = input("새 내용 : ")
    a[4] = today.strftime('%Y-%m-%d %H:%M:%S')
    print(f"{cmd_bit} 번 글이 수정 되었습니다.")

  elif cmd.startswith('article detail'):
    cmd_bit = cmd[14:].strip()

    if cmd_bit == '':
      print("입력된 글 번호가 없습니다.")
      continue

    if not cmd_bit.isdigit():
      print("글 번호가 아닙니다.")
      continue

    article_exist = False
    for i in range(0,len(articles)):
      if articles[i][0] == int(cmd_bit):
        a = articles[i]
        article_exist = True
        break

    if article_exist == False:
      print(f"{cmd_bit} 번 글이 존재하지 않습니다.")
      continue

    print("글 번호 : {}".format(a[0]))
    print("작성 시간 : {}".format(a[3]))
    print("수정 시간 : {}".format(a[4]))
    print("제목 : {}".format(a[1]))
    print("내용 : {}".format(a[2]))

  if cmd == 'exit':
    break

print("== CLI 게시판 종료 ==")
```
