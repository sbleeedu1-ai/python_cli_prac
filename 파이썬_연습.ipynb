```python
article_num = 0
articles = []

print("== CLI 게시판 실행 ==")

while True:
  cmd = input("명령어 ) ")

  if cmd == 'article write':
    article_num += 1
    title = input("제목 : ")
    body = input("내용 : ")
    article = [article_num, title, body]

    articles.append(article)
    print(f"{article_num} 번 글이 생성 되었습니다.")

  elif cmd == 'article list':
    print("=======================================")
    print("번호".ljust(8)+"/",end='')
    print("제목".center(15)+"/",end='')
    print("내용".rjust(8))
    for i in range(len(articles)-1,-1,-1):
      print(f"{articles[i][0]}".ljust(8)+"/",end='')
      print(f"{articles[i][1]}".center(15)+"/",end='')
      print(f"{articles[i][2]}".rjust(8))
    print("=======================================")
  # 거꾸로 출력?
  # articles 의 길이 -1 부터 시작해 0까지 -1 씩 for문

  if cmd == 'exit':
    break

print("== CLI 게시판 종료 ==")
```
